## A website stores user age as a string. Take age input, convert it properly, and decide whether the user should be allowed to register (age ≥ 18).

```python
age_input = input("Enter your age: ")
age = int(age_input)
status_messages = {
    age >= 18: "Registration allowed: You are 18 or older.",
    age < 18: "Registration denied: You must be at least 18 years old."
}

print(status_messages[True])
```

## A temperature monitoring system receives values as text. Convert the input into a number and determine if the temperature is above normal (greater than 37°C).

```python
temp_text = input("Enter the temperature in °C: ")
temp = float(temp_text)
status_messages = {
    temp > 37: "Warning: Temperature is above normal (> 37°C).",
    temp <= 37: "Temperature is normal."
}

print(status_messages[True])
```

## A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
price = float(input("Enter the product price: "))
discount_rate = float(input("Enter the discount rate (in %): "))

discounted_price = price * (1 - discount_rate / 100)

print(f"The discounted price is: {discounted_price}")
```

## You have some apples. Ask the user to enter the total number of apples they have and the number of apples they give to a friend. Write a Python program to calculate how many apples are left.
Which operator will you use to find the remaining apples?

```python
total_apples = int(input("Enter the total number of apples you have: "))
given_apples = int(input("Enter the number of apples given to a friend: "))
apples_left = total_apples - given_apples
print(f"You have {apples_left} apples left.")
```
